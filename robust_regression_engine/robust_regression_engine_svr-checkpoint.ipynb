# PART A — CONCEPTUAL FOUNDATION

##  Q1. What is Regularization in Machine Learning? Why is it needed?

- Regularization is a technique used to reduce overfitting in machine learning.

- It adds a penalty to large model coefficients so that the model becomes simpler
and generalizes better on unseen data.

**Example:**

- If a house price model depends too much on one feature, regularization reduces
that dependency.

## Q2. Difference between Ridge Regression and Lasso Regression



- Ridge Regression uses L2 regularization. It reduces large coefficients but
usually keeps all features.

- Lasso Regression uses L1 regularization. It can make some coefficients zero,
so it can also perform feature selection.

- Ridge → reduces coefficients
- Lasso → can remove some features

## Q3. What is Cross-Validation and why is it important?


- Cross-validation divides the training data into multiple parts and trains
the model several times.

- It helps us check whether the model performs consistently on different
data subsets and helps reduce the risk of overfitting.

**Example:**

- 5-fold cross-validation trains the model 5 times using different validation
parts.

## Q4. Explain Cross-Validation Techniques

### K-Fold Cross-Validation

- The data is divided into K equal parts. Each part is used once for validation
and the remaining parts are used for training.

### Stratified K-Fold

- It keeps similar target-value groups in each fold. For regression, target
values can first be divided into bins.

### Leave-One-Out Cross-Validation (LOOCV)

- One observation is used for validation and all remaining observations are
used for training. This is repeated for every observation.

### Time Series Split

- Data is divided according to time order. Future data is not used to train
a model for predicting the past.

## Q5. Why are tree-based models less sensitive to feature scaling?

### Q5. Tree-Based Models and Scaling

- Decision trees split data using conditions such as:

area_sqft <= 2000

- They do not calculate distances between feature values.

- Therefore, changing the scale of features usually does not affect tree-based
models.

**Example:**
- Area can be 2000 and location score can be 8.5. The tree can still split
them using thresholds.

# PART D — CROSS-VALIDATION STRATEGIES

## Q13. Apply and Compare CV Techniques

```python
ridge_cv_model = Pipeline([
    ("scaler", StandardScaler()),
    ("model", Ridge(alpha=1.0))
])
```

```python
# k fold

kf = KFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

kfold_scores = cross_val_score(
    ridge_cv_model,
    X_train,
    y_train,
    cv=kf,
    scoring="neg_root_mean_squared_error"
)

print("K-Fold RMSE scores:")
print(-kfold_scores)

print("Mean RMSE:", -kfold_scores.mean())
```

## 13.2 Stratified K-Fold Cross-Validation

```python
y_bins = pd.qcut(
    y_train,
    q=5,
    labels=False,
    duplicates="drop"
)

stratified_kf = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)
```

```python
stratified_scores = []

for train_idx, val_idx in stratified_kf.split(X_train, y_bins):
    
    X_tr = X_train.iloc[train_idx]
    X_val = X_train.iloc[val_idx]
    
    y_tr = y_train.iloc[train_idx]
    y_val = y_train.iloc[val_idx]
    
    ridge_cv_model.fit(X_tr, y_tr)
    
    pred = ridge_cv_model.predict(X_val)
    
    rmse = np.sqrt(mean_squared_error(y_val, pred))
    
    stratified_scores.append(rmse)

print("Stratified K-Fold RMSE:")
print(stratified_scores)

print("Mean RMSE:", np.mean(stratified_scores))
```

## 13.3 Leave-One-Out Cross-Validation — LOOCV

```python
loocv_data = X_train.iloc[:300]
loocv_target = y_train.iloc[:300]
```

```python
loo = LeaveOneOut()

loo_errors = []

for train_idx, val_idx in loo.split(loocv_data):
    
    X_tr = loocv_data.iloc[train_idx]
    X_val = loocv_data.iloc[val_idx]
    
    y_tr = loocv_target.iloc[train_idx]
    y_val = loocv_target.iloc[val_idx]
    
    ridge_cv_model.fit(X_tr, y_tr)
    
    pred = ridge_cv_model.predict(X_val)
    
    error = (y_val.iloc[0] - pred[0]) ** 2
    
    loo_errors.append(error)

loo_rmse = np.sqrt(np.mean(loo_errors))

print("LOOCV RMSE:", loo_rmse)
```

### Note:
The complete dataset contains 3800 observations. Full LOOCV requires
3800 model training operations, so a 300-row sample is used for practical
execution time.

## 13.4 Time Series Split

```python
df_time = df.sort_values("sale_date").reset_index(drop=True)
```

```python
X_time = df_time[features]
y_time = df_time["house_price_inr"]
```

```python
# Time Series Split
tscv = TimeSeriesSplit(n_splits=5)

time_scores = []

for train_idx, val_idx in tscv.split(X_time):
    
    X_tr = X_time.iloc[train_idx]
    X_val = X_time.iloc[val_idx]
    
    y_tr = y_time.iloc[train_idx]
    y_val = y_time.iloc[val_idx]
    
    ridge_cv_model.fit(X_tr, y_tr)
    
    pred = ridge_cv_model.predict(X_val)
    
    rmse = np.sqrt(mean_squared_error(y_val, pred))
    
    time_scores.append(rmse)

print("Time Series RMSE:")
print(time_scores)

print("Mean RMSE:", np.mean(time_scores))
```

## Q14. Analyze Performance Metrics Across CV Strategies

```python
cv_results = pd.DataFrame({
    "CV_Method": [
        "K-Fold",
        "Stratified K-Fold",
        "LOOCV (300 sample)",
        "Time Series Split"
    ],
    
    "Mean_RMSE": [
        -kfold_scores.mean(),
        np.mean(stratified_scores),
        loo_rmse,
        np.mean(time_scores)
    ]
})

cv_results
```

```python
plt.figure(figsize=(9,5))

plt.bar(
    cv_results["CV_Method"],
    cv_results["Mean_RMSE"]
)

plt.xticks(rotation=20)
plt.ylabel("RMSE")
plt.title("Comparison of Cross-Validation Strategies")
plt.tight_layout()
plt.show()
```

### Interpretation

Lower RMSE means better prediction error.

Different cross-validation strategies can produce different RMSE values
because each strategy creates training and validation data differently.

Time Series Split is useful when future house prices should be predicted
using past information.

## Q24. Identify Signs of Overfitting or Underfitting

```python
overfit_results = []

models = {
    "Ridge": best_ridge,
    "Lasso": best_lasso,
    "Decision Tree": best_tree,
    "Random Forest": random_forest,
    "SVR RBF": svr_rbf
}

for name, model in models.items():
    
    train_pred = model.predict(X_train)
    test_pred = model.predict(X_test)
    
    train_rmse = np.sqrt(
        mean_squared_error(y_train, train_pred)
    )
    
    test_rmse = np.sqrt(
        mean_squared_error(y_test, test_pred)
    )
    
    train_r2 = r2_score(y_train, train_pred)
    test_r2 = r2_score(y_test, test_pred)
    
    overfit_results.append([
        name,
        train_rmse,
        test_rmse,
        train_r2,
        test_r2
    ])

overfit_df = pd.DataFrame(
    overfit_results,
    columns=[
        "Model",
        "Train_RMSE",
        "Test_RMSE",
        "Train_R2",
        "Test_R2"
    ]
)

overfit_df
```

### Overfitting

If training performance is much better than testing performance,
the model may be overfitting.

For example:

Very low Train RMSE + much higher Test RMSE
→ possible overfitting.

Low Train R² + low Test R²
→ possible underfitting.

Similar Train and Test performance
→ better generalization.
